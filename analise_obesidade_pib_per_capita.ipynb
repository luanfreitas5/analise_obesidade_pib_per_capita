```python
df_gdp = pd.read_csv("data/GDP.csv", decimal=".")
```

```python
df_gdp["Year"] = df_gdp["Year"].apply(lambda x: int(x.split("/")[-1]))
```

- Preecha os anos ausentes em cada país com uma estimativa, baseada na diferença entre o próximo registro e o anterior.


```python
df_gdp.head()
```

```python
arr_years = np.arange(df_gdp["Year"].min(), df_gdp["Year"].max())
df_all_years = pd.DataFrame(arr_years, columns=["Year"])
df_all_years.index = df_all_years["Year"]

df_years_off = ~df_all_years["Year"].isin(df_gdp["Year"])
df_years_off = df_years_off[df_years_off].index
```

```python
df_gdp = df_gdp.sort_values(["Country", "Year"])

df_gdp["delta_gdp"] = df_gdp["gdp_pp"] - df_gdp["gdp_pp"].shift(1)
df_gdp["delta_year"] = df_gdp["Year"] - df_gdp["Year"].shift(1)
df_gdp["gdp_year"] = (df_gdp["delta_gdp"] / df_gdp["delta_year"]).shift(-1)
```

```python
df_gdp["next_year"] = df_gdp["Year"].shift(-1)
del df_gdp["delta_gdp"], df_gdp["delta_year"]
```

```python
df_new_data = pd.DataFrame()

for idx, row in df_gdp.iterrows():
    if row["Year"] == 2011:
        continue

    # print(idx, row)
    year_to_add = df_years_off[
        (df_years_off < row["next_year"]) & (df_years_off > row["Year"])
    ]

    for new_year in year_to_add:
        add_row = row.copy()
        add_row["gdp_pp"] = (new_year - add_row["Year"]) * add_row[
            "gdp_year"
        ] + add_row["gdp_pp"]
        add_row["Year"] = new_year
        df_new_data["kind"] = "estimated"
        df_new_data = pd.concat([df_new_data, add_row.to_frame().transpose()])
```

```python
df_gdp.iloc[100:].head(10)
```

```python
df_gdp = pd.concat([df_gdp, df_new_data]).sort_values(["Country", "Year"])
df_gdp["kind"].fillna("real", inplace=True)
df_gdp.index = df_gdp["Year"]
```

```python
fig, ax = plt.subplots(figsize=(20, 5))
df_gdp[(df_gdp["kind"] == "real") & (df_gdp["Country"] == "United States")].plot(
    kind="scatter", y="gdp_pp", x="Year", ax=ax
)
df_gdp[(df_gdp["kind"] == "estimated") & (df_gdp["Country"] == "United States")].plot(
    kind="scatter", y="gdp_pp", x="Year", ax=ax, color="red"
)
```

### Analises da evolução do PIB per capita dos Estados Unidos

O gráfico mostra a evolução do PIB per capita dos Estados Unidos ao longo dos anos, diferenciando valores reais (em azul) e estimados (em vermelho) para os anos ausentes. Observa-se uma tendência geral de crescimento do PIB per capita, com aumentos mais acentuados em determinados períodos. Os pontos estimados acompanham a tendência dos dados reais, preenchendo lacunas e mantendo a coerência do padrão de crescimento ao longo do tempo. Não há quedas prolongadas, indicando um avanço econômico consistente no período analisado.

- Há uma relação entre níveis de obesidade e gdp per capita?


```python
df_obesity["country-year"] = (
    df_obesity["Country"]
    + "-"
    + df_obesity.reset_index()["Year"].apply(lambda x: str(int(x))).values
)
dict_obesity_year = df_obesity.set_index("country-year").to_dict()["Obesity"]
```

```python
df_gdp["country-year"] = (
    df_gdp["Country"] + "-" + df_gdp["Year"].apply(lambda x: str(x)).values
)
df_gdp["Obesity"] = df_gdp["country-year"].map(dict_obesity_year)
```

```python
df_gdp_clean = df_gdp.dropna()
```

```python
countries = df_gdp_clean["Country"].unique()
```

```python
corr_list = []
for country in countries:
    corr_list += [
        df_gdp_clean[df_gdp_clean["Country"] == country][["Obesity", "gdp_pp"]]
        .corr()
        .iloc[0, 1]
    ]

df_corr = pd.Series(corr_list, index=countries)
```
